### 1.1 Gerando uma Série Temporal Simples



#### Teoria

##### Série Temporal Simples

A série temporal mais simples é uma onda senoidal, que pode ser representada matematicamente como $ y(t) = A \sin(2\pi f t + \phi) $.



#### Prática

###### Código para gerar uma série temporal simples

Usando Python e a biblioteca NumPy para gerar uma série temporal.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate a simple sine wave
np.random.seed(0)
n_points = 1000
x = np.linspace(0, 50, n_points)
y = np.sin(x)

plt.plot(x, y)
plt.title("Série Temporal Simples")
plt.xlabel("Tempo")
plt.ylabel("Valor")
plt.show()
```

### 1.2 Adicionando Ruído



#### Teoria

##### O que é ruído?

Ruído é uma variação aleatória nos dados que pode ser causada por diversos fatores.



#### Prática


```python
# Add some noise
y_noise = y + np.random.normal(0, 0.1, 1000)
plt.plot(x, y_noise)
plt.title("Série Temporal com Ruído")
plt.xlabel("Tempo")
plt.ylabel("Valor")
plt.show()
```

### 1.3 Introduzindo Anomalias





#### Teoria

##### O que são anomalias?

Pontos de dados que se desviam significativamente do restante da série.

#### Prática

```python
# Introduce anomalies
anomalies = np.array([3*n_points/10, 5*n_points/10, 7*n_points/10, 8*n_points/10]).astype(int)
y_noise[anomalies] = y_noise[anomalies] + 2

plt.plot(x, y_noise)
plt.scatter(x[anomalies], y_noise[anomalies], color='red')
plt.title("Série Temporal com Anomalias")
plt.xlabel("Tempo")
plt.ylabel("Valor")
plt.show()
```

Agora, vamos adicionar a implementação dos métodos para dividir os dados e adicionar anomalias.

```python
import numpy as np
import matplotlib.pyplot as plt

class TimeSeries:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def generate_time_series(n_points=1000, noise_std=0.1, seed=None):
        """
        Generate a noisy sine wave time series with anomalies.
    
        Parameters:
        - n_points: Number of data points
        - noise_std: Standard deviation of the Gaussian noise
        - seed: Random seed for reproducibility
    
        Returns:
        - x: Time steps
        - y_noise: Noisy time series with anomalies
        """

        if seed is not None:
            np.random.seed(seed)
        
        # Generate the time series
        x = np.linspace(0, 100, n_points)
        y = np.sin(x)
        y_noise = y + np.random.normal(0, noise_std, n_points)
    
        return x, y_noise
    
    def split_time_series(x, y, train_ratio=0.6, val_ratio=0.2):
        """
        Split the time series data into training, validation, and test sets.
    
        Parameters:
        - x: Time steps
        - y: Time series values
        - train_ratio: Proportion of data for training
        - val_ratio: Proportion of data for validation
    
        Returns:
        - x_train, y_train: Training set
        - x_val, y_val: Validation set
        - x_test, y_test: Test set
        """
        total_points = len(x)
        train_size = int(train_ratio * total_points)
        val_size = int(val_ratio * total_points)
    
        x_train, x_val, x_test = x[:train_size], x[train_size:train_size+val_size], x[train_size+val_size:]
        y_train, y_val, y_test = y[:train_size], y[train_size:train_size+val_size], y[train_size+val_size:]
        return x_train, y_train, x_val, y_val, x_test, y_test
    
        return x_train, y_train, x_val, y_val, x_test, y_test

    def add_global_anomalies(data_dict, anomaly_percentage=0.05, anomaly_start_percentage=0.1, 
                          train_anomaly_ratio=0.5, val_anomaly_ratio=0.3, anomaly_magnitude=3, seed=None):
        """
        Add anomalies to multiple time series data based on the total count across all series.
    
        Parameters:
        - data_dict: Dictionary containing the time series data divided into 'train', 'val', and 'test'
        - anomaly_percentage: Percentage of total points across all series that should be anomalies
        - anomaly_start_percentage: Percentage point in each series where anomalies should start
        - train_anomaly_ratio: Proportion of anomalies in the training set
        - val_anomaly_ratio: Proportion of anomalies in the validation set
        - anomaly_magnitude: Magnitude of the anomalies
        - seed: Random seed for reproducibility
    
        Returns:
        - updated_data_dict: Dictionary containing the time series with added anomalies
        - anomaly_indices_dict: Dictionary containing the indices where anomalies were introduced in each series
        """
        if seed is not None:
            np.random.seed(seed)

        # Validate the data_dict structure
        if not all(key in data_dict for key in ['train', 'val', 'test']):
            raise ValueError("data_dict must contain 'train', 'val', and 'test' keys.")
        
        if train_anomaly_ratio + val_anomaly_ratio >= 1:
            raise ValueError("The sum of train_anomaly_ratio and val_anomaly_ratio should not exceed 1.")
        
        # Calculate total points in dataset and total anomalies for generation
        total_points = sum(len(series) for series in data_dict.values())
        # Select global anomalies starting from the anomaly_start_index
        anomaly_start_index = int(anomaly_start_percentage * total_points)
        # Calculate the total number of anomalies to generate
        total_anomalies = int(anomaly_percentage * total_points)

        if total_anomalies <= 0:
            raise ValueError("Total number of anomalies calculated as zero or negative. Increase the anomaly_percentage.")
    
        # Determine how many anomalies go into each data split
        train_anomalies = int(train_anomaly_ratio * total_anomalies)
        val_anomalies = int(val_anomaly_ratio * total_anomalies)
        test_anomalies = total_anomalies - train_anomalies - val_anomalies
    
        updated_data_dict = {}
        anomaly_indices_dict = {}

        # Assign anomalies to each data partition ('train', 'val', 'test')
        for key in ['train', 'val', 'test']:
            # Carrega a série de dados correspondente à chave atual ('train', 'val' ou 'test').
            series = data_dict[key]
            series_length = len(series)
            # Determina o número de anomalias que devem ser adicionadas à série atual com base na parte do conjunto de dados que ela representa.
            subset_anomalies = train_anomalies if key == 'train' else val_anomalies if key == 'val' else test_anomalies

            # if key == 'train':
            #     subset_anomalies = train_anomalies
            # elif key == 'val':
            #     subset_anomalies = val_anomalies
            # else:
            #     subset_anomalies = test_anomalies

            if  subset_anomalies > 0:
                # Generate anomaly indices directly within the segment
                start_index = anomaly_start_index if key == 'train' else 0
                local_anomaly_indices = np.random.choice(range(start_index, series_length), subset_anomalies, replace=False)
                series_anomalous = np.copy(series)
                series_anomalous[local_anomaly_indices] += anomaly_magnitude

                updated_data_dict[key] = series_anomalous
                anomaly_indices_dict[key] = local_anomaly_indices
            else:
                updated_data_dict[key] = np.copy(series)
                anomaly_indices_dict[key] = []
            
    
        return updated_data_dict, anomaly_indices_dict
    

    # Função para criar janelas de dados com passo específico
    def create_windows_with_step(data, window_size, step):
        windows = []
        for i in range(0, len(data) - window_size + 1, step):
            windows.append(data[i:i+window_size])
        return np.array(windows)

    def create_windows_and_labels(data, labels, window_size, step):
        windows = []
        window_labels = []
        for i in range(0, len(data) - window_size + 1, step):
            windows.append(data[i:i+window_size])
            window_labels.append(1 if np.any(labels[i:i+window_size]) else 0)
        return np.array(windows), np.array(window_labels)

    def create_windows_and_sample_labels(data, labels, window_size, step):
        windows = []
        sample_labels = []
        for i in range(0, len(data) - window_size + 1, step):
            windows.append(data[i:i+window_size])
            sample_labels.append(labels[i:i+window_size])
        return np.array(windows), np.array(sample_labels)
    
```

```python
from sklearn.model_selection import GridSearchCV
from tensorflow.keras.wrappers.scikit_learn import KerasClassifier
from tensorflow.keras.callbacks import EarlyStopping
import tensorflow as tf
import matplotlib.pyplot as plt

def weighted_binary_crossentropy2(pos_weight=1, neg_weight=1):
    def loss(y_true, y_pred):
        loss = - (pos_weight * y_true * tf.math.log(y_pred + 1e-5) + neg_weight * (1 - y_true) * tf.math.log(1 - y_pred + 1e-5))
        return tf.reduce_mean(loss)
    return loss

def weighted_binary_crossentropy(pos_weight=1, neg_weight=1):
    def loss(y_true, y_pred):
        y_true = tf.cast(y_true, tf.float32)  # Converter y_true para float32
        loss = - (pos_weight * y_true * tf.math.log(y_pred + 1e-5) + neg_weight * (1 - y_true) * tf.math.log(1 - y_pred + 1e-5))
        return tf.reduce_mean(loss)
    return loss

def focal_loss(gamma=2., alpha=4.):
    def loss(y_true, y_pred):
        pt_1 = tf.where(tf.equal(y_true, 1), y_pred, tf.ones_like(y_pred))
        pt_0 = tf.where(tf.equal(y_true, 0), y_pred, tf.zeros_like(y_pred))
        return -tf.reduce_sum(alpha * tf.pow(1. - pt_1, gamma) * tf.math.log(pt_1 + 1e-5)) \
               -tf.reduce_sum((1 - alpha) * tf.pow(pt_0, gamma) * tf.math.log(1. - pt_0 + 1e-5))
    return loss

def auc_loss(y_true, y_pred):
    auc = tf.metrics.AUC(y_true, y_pred)[1]
    return 1 - auc

class PrintLogs(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        print(f"Epoch: {epoch + 1}, Loss: {logs['loss']:.4f}, Val Loss: {logs['val_loss']:.4f}, Accuracy: {logs['accuracy']:.4f}, Val Accuracy: {logs['val_accuracy']:.4f}")

# Definir modelo como uma função
def create_model(lstm_units, loss, optimizer='adam', activation='sigmoid'):
    # Definir o otimizador, a função de ativação e a função de perda
    if loss == 'weighted_binary_crossentropy':
        loss = weighted_binary_crossentropy(pos_weight=1, neg_weight=1)
    if loss == 'weighted_binary_crossentropy2':
        loss = weighted_binary_crossentropy2(pos_weight=1, neg_weight=1)
    elif loss == 'focal_loss':
        loss = focal_loss(gamma=1., alpha=.5)
    elif loss == 'auc_loss':
        loss = auc_loss

    model = tf.keras.Sequential([
        tf.keras.layers.LSTM(lstm_units, input_shape=(hiperparameter_window_size, 1), return_sequences=True),
        tf.keras.layers.LSTM(25),
        tf.keras.layers.Dense(1, activation=activation)
    ])
    model.compile(optimizer=optimizer, loss=loss, metrics=['accuracy'])
    
    return model

# Definir o espaço de hiperparâmetros
param_grid = {
    #'window_size': [hiperparameter_window_size, pow(hiperparameter_window_size, 2), pow(hiperparameter_window_size, 4)],
    #'step_size': [hiperparameter_window_size/4, hiperparameter_window_size/2, hiperparameter_window_size],
    #'batch_size': [16, 32, 64],
    #'lstm_units': [hiperparameter_window_size, pow(hiperparameter_window_size, 2), pow(hiperparameter_window_size, 3)],
    'batch_size': [4],
    'lstm_units': [hiperparameter_window_size],
    #'loss': ['binary_crossentropy', 'weighted_binary_crossentropy2', 'focal_loss', 'auc_loss'],
    'loss': ['binary_crossentropy', 'weighted_binary_crossentropy2'],
}


print(tf.__version__)
print("GPUs disponíveis: ", len(tf.config.experimental.list_physical_devices('GPU')))
# 1. Preparar os Dados
# Suponhamos que 'windows_train_1' e 'window_labels_train_1' são seus dados de treinamento
# e 'windows_val_1' e 'window_labels_val_1' são seus dados de validação
train_data = np.expand_dims(windows_train_1, axis=-1)
val_data = np.expand_dims(windows_val_1, axis=-1)


# Criar o modelo Keras com o Scikit-Learn wrapper
class KerasClassifierWithCallbacks(KerasClassifier):
    def fit(self, x, y, **kwargs):
        model = self.build_fn(**self.filter_sk_params(self.build_fn))
        print_logs = PrintLogs()
        early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
        return super().fit(x, y, validation_data=(val_data, window_labels_val_1), callbacks=[early_stopping], **kwargs)

model = KerasClassifierWithCallbacks(build_fn=create_model, epochs=50, verbose=0)

# Realizar a pesquisa em grade
grid = GridSearchCV(estimator=model, param_grid=param_grid, n_jobs=2, cv=2, verbose=2, return_train_score=True)
grid_result = grid.fit(train_data, window_labels_train_1)

# Sumarizar os resultados
print("Melhores: %f usando %s" % (grid_result.best_score_, grid_result.best_params_))
```
